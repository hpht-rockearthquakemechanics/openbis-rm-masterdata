```python
import csv
import os

# 1. Lista ordinata con tutte le estensioni .csv rimesse al loro posto
files_to_merge = [
    "BGS_ROCKNAME.csv",
    "MSL_ANALYSIS.csv",
    "MSL_ARTIFICIAL.csv",
    "MSL_ENVIRONMENTCONTROL.csv",
    "MSL_EQUIPMENT.csv",
    "MSL_FACILITY.csv",
    "MSL_FACILITYTYPE.csv",
    "MSL_GEOLOGICALAGE.csv",
    "MSL_INFERRED.csv",
    "MSL_MEASUREDPROPERTY.csv",
    "MSL_MICROSTRUCTURE.csv",   
    "MSL_MINERAL.csv",          
    "MSL_MODEL.csv",            
    "MSL_POREFLUID.csv",        
    "MSL_ROCK.csv",             
    "MSL_SEDIMENT.csv",         
    "MSL_SETTING.csv",          
    "MSL_SOFTWARE.csv",         
    "QUDT_QUANTITYKIND.csv",    
    "QUDT_UNIT.csv",       
    "SWEET_MATR_IND.csv",       
    "SWEET_STATE_PHYS.csv"
]

output_merged_csv = "external_dictionaries_merged.csv"

print("Avvio della pipeline ULTRA-RAPIDA di unione (Lettura: , | Scrittura: ;)")
print("=" * 80)

processed_count = 0
total_rows_written = 0

# Apertura del file master in scrittura progressiva con separatore PUNTO E VIRGOLA (;)
with open(output_merged_csv, 'w', encoding='utf-8-sig', newline='') as f_out:
    writer = csv.writer(f_out, delimiter=';')
    
    for filename in files_to_merge:
        if not os.path.exists(filename):
            print(f"   [AVVISO] File '{filename}' non trovato in locale. Salto questo blocco.")
            continue
            
        print(f"-> Integrazione e conversione per: '{filename}'...")
        
        # Inserisce una riga vuota di separazione tra un dizionario e l'altro
        if processed_count > 0:
            writer.writerow([])
            total_rows_written += 1
            
        with open(filename, 'r', encoding='utf-8-sig', errors='ignore') as f_in:
            # LETTURA: I file di input sono separati da VIRGOLA (,)
            reader = csv.reader(f_in, delimiter=',')
            
            # Leggiamo le prime 4 righe per estrarre i metadati strutturali
            try:
                r0 = next(reader)
                r1 = next(reader)
                r2 = next(reader)
                r3 = next(reader)
            except StopIteration:
                print(f"   [SALTO] Il file '{filename}' ha meno di 4 righe ed è corrotto.")
                continue
                
            # Scrittura dei metadati OpenBIS riformattati a 4 colonne (scrive con ;)
            writer.writerow(["VOCABULARY_TYPE", "", "", ""])
            writer.writerow(["Version", "Code", "Description", ""])
            
            # Recuperiamo i codici corretti dalle posizioni della virgola
            vocab_code = r2[0].strip() if len(r2) > 0 else ""
            vocab_desc = r2[1].strip() if len(r2) > 1 else ""
            writer.writerow(["5", vocab_code, vocab_desc, ""])
            
            writer.writerow(["Version", "Code", "Label","Description"])
            total_rows_written += 4
            
            # --- CICLO SUI DATI ATOMICI ---
            for row in reader:
                if not row:
                    continue
                
                # Ora le colonne sono divise correttamente grazie al separatore a virgola
                code_val = row[0].strip() if len(row) > 0 else ""
                desc_val = row[1].strip() if len(row) > 1 else ""
                label_val = row[2].strip() if len(row) > 2 else ""
                
                # Salto delle righe vuote o spurie di Excel
                if not code_val and not desc_val and not label_val:
                    continue
                    
                # Scrittura su disco: inserisce il '5' e separa i campi con il punto e virgola (;)
                writer.writerow(["5", code_val, label_val, desc_val])
                total_rows_written += 1
                
        processed_count += 1

print("=" * 80)
print(f"COMPLETATO CON SUCCESSO! Generato il file unico: '{output_merged_csv}'")
print(f"Totale file fusi: {processed_count} su {len(files_to_merge)}")
print(f"I campi sono ora separati nativamente da ';' e privi di macro-stringhe compresse.")
print("=" * 80)
```
